```python
%%sql plan_prep <<
# Cleaning of the plan entity
with plan_prep as (
    select
        pe.plan_id,

        -- Project details
        pe.project_id__hashed,
        p.project_type,
        p.organization_name,
        p.device_type,

        -- Plan details
        pe.plan_created_at,
        pe.plan_name,
        pe.network_provider_id,
        pe.plan_price_amount_local,
        pe.data_allowance_mb,
        pe.is_unlimited_data,
        pe.voice_allowance_seconds,
        pe.is_unlimited_voice,
        pe.sms_allowance,
        pe.is_unlimited_sms,
        pe.validity_value,
        pe.validity_unit

    from plan_events as pe
    left join projects as p
        on p.project_id__hashed = pe.project_id__hashed
    where _is_current_state
)
select *
from plan_prep
```

```python
%%sql usage_prep <<
# Cleaning of the subscription log.
with usage_prep as (
    select
        subscription_id,
        plan_id as u_plan_id,
        reporting_date,
        date_trunc('month', reporting_date) as reporting_month,
        subscription_period_start,
        subscription_period_end,
        subscription_period_number,
        cumulative_data_usage_megabyte,
        cumulative_voice_usage_minutes,
        cumulative_sms_usage,
        number_of_addons_activated
    from usage_data
    qualify row_number() over (partition by subscription_id, subscription_period_number order by subscription_period_end desc) = 1
)
select *
from usage_prep
```

```python
%%sql df_base <<
# Creation of wide-table for general usage.
select
    up.*,
    pp.*

from usage_prep as up
left join plan_prep as pp
    on pp.plan_id = up.u_plan_id
```

## Q3: Compare the retention pattern for the most recently launched project versus the two older ones.

```python
%%sql df_q3 <<
# Foundation model keeps a row for subsciptions even after they are inactive, allowing for the calculation of churn.
with date_base_prep as (
    select
        reporting_month
    from df_base
    where reporting_month between '2024-05-01' and '2025-05-01'
    group by 1
),
subscriptions_prep as (
    select
        subscription_id,
        organization_name,
        min(reporting_month) as first_sub_at,
        max(reporting_month) as last_sub_at 
            /* 
                DISCLAIMER: this logic does not consider the few reactivated subscriptions churned during the gap.
                Ideally I would use lag(reporting_month) ignore nulls over (...) at the later CTE to get the last non null value allowing to capture the proper delta between current date and inactivity.
                Unfortunatley the ingore null does not work in the notebook (not sure what SQL syntax it is using exactly), using this as a stopgap.
            */
    from df_base
    where reporting_month between '2024-05-01' and '2025-05-01'
    group by 1,2
),
foundation as (
    select
        dbp.reporting_month,
        sp.subscription_id,
        sp.organization_name
    from date_base_prep as dbp
    left join subscriptions_prep as sp
        on dbp.reporting_month >= sp.first_sub_at
        and datediff('month', sp.last_sub_at, dbp.reporting_month) <= 1 # considering subscription churned for 1 month. The exact number needs to be determined with business based on factors like potential reactivation dates
)
select
    f.subscription_id as base_id,
    f.reporting_month as base_month,
    f.organization_name as base_org,
    case when df.subscription_id is not null and df.subscription_period_number = 1 then 'new'
        when df.subscription_id is not null then 'retained'
    else 'churned' end as sub_category_churn,
    df.*
from foundation as f
left join df_base as df
    on df.reporting_month = f.reporting_month
    and df.subscription_id = f.subscription_id
```

```python
%%sql df_q3b <<
# Aggregation and cuts of foundation model.
select
    base_month,
    sub_category_churn,
    count(distinct base_id) as sub_total,
    sum(sub_total) over (partition by base_month) as month_total,
    sub_total / if(month_total is null, 0, month_total)*100 as month_share
from df_q3
where 1=1
--and base_org = 'ACME Phone'
--and base_org = 'SmartDevices Inc.'
and base_org = 'People Mobile'
group by 1,2
order by 1 desc, 2 desc
```

```python
fig1 = px.bar(df_q3b, x="base_month", y="sub_total", color="sub_category_churn", text_auto=True)
fig2 = px.area(df_q3b, x="base_month", y="month_share", color="sub_category_churn", markers=True)
fig1.show()
fig2.show()
```
